```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
test_df = pd.read_csv('/content/test.csv')
train_df = pd.read_csv('/content/train.csv')
```

# EDA AND DATA CLEANING

```python
train_df.info()
```

```python

train_df['HasCabin'] = train_df['Cabin'].notnull().astype(int)
test_df['HasCabin'] = test_df['Cabin'].notnull().astype(int)
```

```python
train_df = train_df.drop(columns='Cabin',axis=1)
test_df = test_df.drop(columns='Cabin',axis=1)
```

```python
train_df.value_counts()
```

```python
print(train_df.corr(numeric_only=True)["Survived"])
```

```python
train_df = train_df.drop(columns='PassengerId',axis=1)
test_df = test_df.drop(columns='PassengerId',axis=1)
```

```python
train_df.info()
```

```python
train_df["Age"].fillna(train_df["Age"].median(), inplace=True)
test_df["Age"].fillna(test_df["Age"].median(), inplace=True)

train_df
```

```python
train_df["Embarked"].fillna(
    train_df["Embarked"].mode()[0],
    inplace=True
)
test_df["Embarked"].fillna(
    test_df["Embarked"].mode()[0],
    inplace=True
)
```

```python
train_df.info()
```

```python
print(train_df.corr(numeric_only=True)["Survived"])
```

```python
print(train_df["Ticket"].nunique())
print(train_df["Ticket"].count())
```

```python
print(train_df.corr(numeric_only=True)["Survived"])
```

```python
train_df["FamilySize"] = (
    train_df["SibSp"] +
    train_df["Parch"] + 1
)
test_df["FamilySize"] = (
    test_df["SibSp"] +
    test_df["Parch"] + 1
)
train_df
```

```python
print(train_df.corr(numeric_only=True)["Survived"])
```

```python
train_df.groupby("FamilySize")["Survived"].mean()
```

```python
train_df["IsAlone"] = (
    train_df["FamilySize"] == 1
).astype(int)

test_df["IsAlone"] = (
    test_df["FamilySize"] == 1
).astype(int)
```

```python
train_df
```

```python
train_df['Title'] = train_df['Name'].str.extract(r' ([A-Za-z]+)\.', expand=False)

test_df['Title'] = test_df['Name'].str.extract(r' ([A-Za-z]+)\.', expand=False)
```

```python
train_df
```

```python
train_df.drop("Name", axis=1, inplace=True)
test_df.drop("Name", axis=1, inplace=True)
```

```python
train_df.drop(["SibSp", "Parch"], axis=1, inplace=True)

test_df.drop(["SibSp", "Parch"], axis=1, inplace=True)
```

```python
train_df.drop("Ticket", axis=1, inplace=True)
test_df.drop("Ticket", axis=1, inplace=True)
```

```python
train_df
```

```python
cols = ['Sex','Embarked','Title']
train_df = pd.get_dummies(train_df, columns=cols)
test_df = pd.get_dummies(test_df, columns=cols)
```

```python
train_df
```

```python
train_df = train_df.astype(int)
train_df
```

```python
test_df.info()
```

```python
test_df['Fare'].fillna(test_df['Fare'].median(),inplace=True)
```

```python
test_df.info()
```

```python
test_df = test_df.astype(int)
test_df
```

# Feature Scaling

```python
X = train_df.drop(columns='Survived')
y = train_df['Survived']
```

```python
X
```

```python
X.value_counts()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

cols = ['Pclass', 'Age', 'Fare', 'FamilySize']

X_scaled = X.copy()
test_df_scaled = test_df.copy()

X_scaled[cols] = scaler.fit_transform(X[cols])

test_df_scaled[cols] = scaler.transform(test_df[cols])
X_scaled
```

# Models Scaled

```python
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import GridSearchCV
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
models = {
    "KNN": (
        KNeighborsClassifier(),
        {
            "n_neighbors": [3, 5, 7, 9, 11],
            "weights": ["uniform", "distance"],
            "metric": ["euclidean", "manhattan"]
        }
    ),

    "Logistic Regression": (
        LogisticRegression(max_iter=1000),
        {
            "C": [0.01, 0.1, 1, 10, 100],
            "solver": ["liblinear", "lbfgs"],
            "penalty": ["l2"]
        }
    ),
    "SVM": (
    SVC(),
    {
        "C": [0.1, 1, 10],
        "kernel": ["linear", "rbf"]
    }
)
}

base_normal = {}

for model_name, (model, params) in models.items():

    grid = GridSearchCV(
        model,
        params,
        cv=5,
        scoring='accuracy'
    )

    grid.fit(X_scaled, y)

    print(f"\n{model_name}")

    base_normal[model_name] = grid.best_estimator_

    print("Best Params:", grid.best_params_)

    print("Best Accuracy:", grid.best_score_)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

models = {
        "Decision Tree": (
          DecisionTreeClassifier(random_state=42),
          {
              "criterion": ["gini", "entropy"],
              "max_depth": [3, 5, 7, 10],
              "min_samples_split": [2, 5, 10]
          }
        ),
        "Random Forest": (
    RandomForestClassifier(random_state=42),
    {
    "n_estimators": [50, 100],
    "max_depth": [3, 5],
    "min_samples_split": [2, 5]
}
),
        "XGBoost": (
    XGBClassifier(
        eval_metric='logloss',
        random_state=42
    ),
    {
    "n_estimators": [50, 100],
    "max_depth": [2, 3, 4],
    "learning_rate": [0.05, 0.1],
}
)
}

best_models = {}


for model_name, (model, params) in models.items():
      grid = GridSearchCV(
            model,
            params,
            cv=5,
            scoring='accuracy'
        )

      grid.fit(X, y)

      print(f"\n{model_name}")
      best_models[model_name] = grid.best_estimator_

      print("Best Params:", grid.best_params_)

      print("Best Accuracy:", grid.best_score_)
```

```python
original_test_df = pd.read_csv("/content/test.csv")

passenger_ids = original_test_df["PassengerId"]

best_rf = best_models["Random Forest"]

predictions = best_rf.predict(test_df)

submission = pd.DataFrame({
    "PassengerId": passenger_ids,
    "Survived": predictions
})
```

```python
submission.to_csv("submission.csv", index=False)
```
